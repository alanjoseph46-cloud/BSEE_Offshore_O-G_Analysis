## 1. Dataset 1 — Production by Platform
* **Granularity:** Platform $\times$ Production Month  
* **Description:** Monthly operational and fluid production performance metrics for offshore platform structures.

| Column Name | Data Type | Description |
| :--- | :--- | :--- |
| `structure_number` | Identifier | Unique platform/structure registration ID. Primary link to Dataset 2. |
| `structure_name` | String | Commercial or administrative name assigned to the offshore platform. |
| `production_month` | Date (`YYYY-MM-01`) | Reporting month for the recorded production volume rates. |
| `producing_wells` | Integer | Total count of active producing wells connected to the platform during the month. |
| `oil_rate_bpd` | Float | Average daily oil production rate in barrels per day (BPD). |
| `gas_rate_mcfpd` | Float | Average daily natural gas production rate in thousand cubic feet per day (MCFPD). |
| `oil_equivalent_rate_bpd` | Float | Combined energy rate in barrels of oil equivalent per day (BOEPD). |
| `water_rate_bpd` | Float | Average daily produced water volume rate in barrels per day (BPD). |
| `operator_name` | String | Operating company managing production activities for the platform. |
| `region` | Categorical | Geographic offshore region/planning area (e.g., Gulf of Mexico). Primary link to Dataset 3. |
| `install_date` | Date | Date when the platform structure was installed on-site. |
| `removal_date` | Date (Nullable) | Date when the platform was decommissioned/removed (`NaN` = Active Platform). |

---

## 2. Dataset 2 — Platform Structures
* **Granularity:** Structure (Unique Platform)  
* **Description:** Master asset registry describing physical characteristics, location, and staffing for each offshore platform.

| Column Name | Data Type | Description |
| :--- | :--- | :--- |
| `structure_number` | Identifier | Unique platform/structure registration ID (Primary Key). |
| `field` | Categorical | Name of the oil/gas field where the platform is installed. |
| `struc_type_code` | Categorical | Structural design classification (e.g., Fixed Leg, Spar, TLP, FPSO). |
| `major_struc_flag` | Boolean / String | Indicator of whether the platform is classified as a major facility (`Y`/`N`). |
| `install_date` | Date | Official installation date recorded in the facility registry. |
| `removal_date` | Date (Nullable) | Official removal/decommissioning date (`NaN` = Active Platform). |
| `water_depth_ft` | Float | Sea depth at the platform location measured in feet. |
| `INCs` | Integer | Historical total Incidents of Non-Compliance (INCs) cited on this specific structure. |
| `latitude` | Float | Geographic coordinate (Y-location in decimal degrees). |
| `longitude` | Float | Geographic coordinate (X-location in decimal degrees). |
| `heliport_flag` | Boolean / String | Presence of a heliport deck on the structure (`Y`/`N`). |
| `attended_8_Hr` | Boolean / String | Indicator if the facility is staffed during standard 8-hour day shifts (`Y`/`N`). |
| `manned_24_Hr` | Boolean / String | Indicator if the facility is continuously manned 24 hours/day (`Y`/`N`). |

---

## 3. Dataset 3 — INC Summary (Safety & Compliance)
* **Granularity:** Region $\times$ Operator  
* **Description:** Aggregated regulatory compliance records and Incidents of Non-Compliance (INCs) summarized by operator and region.

| Column Name | Data Type | Description |
| :--- | :--- | :--- |
| `region` | Categorical | Geographic offshore region/planning area (Composite Key). |
| `operator_name` | String | Operating company responsible for compliance (Composite Key). |
| `warning_incs` | Integer | Count of minor non-compliance enforcement citations issued (Warnings). |
| `component_shut_in_incs` | Integer | Count of moderate violations requiring specific component shutdowns. |
| `facility_shut_in_incs` | Integer | Count of severe violations requiring full facility operational shut-ins. |
| `inspection_records` | Integer | Total number of individual safety inspection records conducted. |
| `facilities_inspected` | Integer | Total number of unique platforms inspected in that region by the operator. |
| `inspection_types` | Integer / String | Diversity or total count of different inspection protocols/types conducted. |

```python
# columns: production_month, install_date,removal_date
# converting the datatype to date. (object-> date)

# using a variable the columns are stored in a list & using for loop the columns are called one by one for datatype conversion.
date_cols_production = ["production_month", "install_date","removal_date"]

for col in date_cols_production:
    production[col] = pd.to_datetime(production[col],errors= "coerce")
    
print("After Correction of columns with date:\n",production[date_cols_production].dtypes)
print("*"*60)

# columns: oil_rate_bpd, gas_rate_mcfpd, oil_equivalent_rate_bpd, water_rate_bpd
# converting the datatype to numerical data type. (object-> numerical)

# using a variable -> columns to  list-> using a for loop -->first remove commas-> then datatype conversion to numeric. 
num_cols_production = ["oil_rate_bpd","gas_rate_mcfpd","oil_equivalent_rate_bpd","water_rate_bpd"]

#But the columns contains "," (COMMAS) ---->so,idea is Object/Text → Remove commas → Numeric
#regex=False- It tells Pandas: Treat "," as a normal character, not as a regular expression.

for col in num_cols_production:
    production[col] = production[col].astype(str).str.replace(",", "",regex=False)  
    production[col] = pd.to_numeric(production[col],errors ="coerce")

print("\nAfter Converting to Numerical datatype\n",production[num_cols_production].dtypes)
```

```python
# Function used to create a DataFrame--> named summary-->create a new summary table
def missing_value_summary(df):
    summary = pd.DataFrame(
        {"missing_value":df.isnull().sum(),
         "missing_percentage_%":df.isnull().mean() * 100
        })

    # Only keep columns with missing values 
    result = summary[summary["missing_value"]>0]
    # sorting 
    return result.sort_values("missing_percentage_%",ascending=False)
```

```python
# Select only meaningful numerical columns
outlier_cols_production = [
    "oil_rate_bpd",
    "gas_rate_mcfpd",
    "oil_equivalent_rate_bpd",
    "water_rate_bpd"
]
```

```python
# Detect outliers using IQR
def outlier_summary(df, columns):
    results = []

    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outlier_mask = (
            (df[col] < lower_bound) |
            (df[col] > upper_bound)
        )

        count = outlier_mask.sum()
        percentage = count / len(df) * 100

        results.append({
            "Column": col,
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Outlier Count": count,
            "Outlier %": round(percentage,2),
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound
        })

    return pd.DataFrame(results)
outlier_summary(production, outlier_cols_production)
```

Key Finding:</h1>
    
we found that in our dataset, over 50% of the rows for rate columns have a value of 0.0.
* This causes two main issues:
    Inflated Outlier Counts:
    * For oil_rate_bpd, Q1 = 0 and Q3 = 136, making IQR = 136.
    * The calculated upper bound becomes 136 + (1.5 \times 136) = 340.
    * Every active production day above 340 bpd will be flagged as an "outlier",categorizing nearly 20% of your valid operational data as anomalies.
    Negative Lower Bounds:
    * Since Q1 = 0, the lower bound evaluates to 0 - (1.5 \times 136) = -204.
    * Physical production rates cannot drop below zero, making the lower bound calculation ineffective.

* To resolve this issue, 
 filter for active production records (> 0) when calculating quartiles, or focus exclusively on the upper bound for non-negative production rates.

  <!-- Overview Statement -->
  <p style="margin: 0 0 14px 0; color: #1E293B; font-size: 14px; line-height: 1.6;">
    Descriptive statistics reveal that <strong>over 50% of records in rate columns contain a value of 0.0</strong>. Applying standard Interquartile Range (IQR) outlier detection to this zero-inflated distribution distorts statistical bounds:
  </p>

  <!-- Issues List -->
  <ul style="margin: 0 0 16px 0; padding-left: 20px; color: #334155; font-size: 14px; line-height: 1.6;">
    
<li style="margin-bottom: 10px;">
      <strong>Inflated Outlier Counts (False Anomaly Flags):</strong><br>
      For <code style="background-color: #E0F2FE; color: #0369A1; padding: 2px 6px; border-radius: 4px;">oil_rate_bpd</code>, $Q_1 = 0$ and $Q_3 = 136$, yielding an $\text{IQR} = 136$. The upper threshold evaluates to:
      <div style="background-color: #FFFFFF; border: 1px solid #CBD5E1; padding: 8px 12px; margin: 6px 0; border-radius: 4px; font-family: monospace; color: #0F172A; font-size: 13px;">
        Upper Bound = Q3 + (1.5 × IQR) = 136 + (1.5 × 136) = 340 bpd
      </div>
      This artificially low cutoff flags <strong>nearly 20% of valid high-volume operational days</strong> as statistical anomalies.
    </li>

<li style="margin-bottom: 0;">
      <strong>Non-Physical Negative Lower Bounds:</strong><br>
      Because $Q_1 = 0$, the standard lower threshold formula yields:
      <div style="background-color: #FFFFFF; border: 1px solid #CBD5E1; padding: 8px 12px; margin: 6px 0; border-radius: 4px; font-family: monospace; color: #0F172A; font-size: 13px;">
        Lower Bound = Q1 - (1.5 × IQR) = 0 - (1.5 × 136) = -204 bpd
      </div>
      Since physical production rates are strictly non-negative ($\ge 0$), this calculation provides no useful filtering mechanism for low-end anomalies.
    </li>

  <!-- Resolution Box -->
  <div style="background-color: #FFFFFF; border: 1px solid #7DD3FC; padding: 12px 16px; border-radius: 6px;">

    <p style="margin: 4px 0 0 0; color: #0F172A; font-size: 13.5px; line-height: 1.5;">
      Filter for <strong>active operational records strictly greater than zero (<code style="background-color: #E0F2FE; color: #0369A1; padding: 1px 5px; border-radius: 3px;">rate &gt; 0</code>)</strong> prior to calculating quartiles and IQR boundaries, or rely exclusively on high-percentile upper thresholds (e.g., 99th percentile) for non-negative production streams.
    </p>
  </div>

## Recommended Remediation Strategy:

```python
#  Calculate Active-Record IQR Outlier Summary
outlier_summary = []

for col in outlier_cols_production:
    # Filter for active records (greater than 0)
    active_data = production[production[col] > 0][col]

    Q1 = active_data.quantile(0.25)
    Q3 = active_data.quantile(0.75)
    IQR = Q3 - Q1

    # Focus on extreme high values
    upper_bound = Q3 + (1.5 * IQR)

    # Count outliers across full dataset
    total_outliers = (production[col] > upper_bound).sum()
    pct_outliers = (total_outliers / len(production)) * 100

    outlier_summary.append(
        {
            "Column": col,
            "Q1 (Active)": round(Q1, 2),
            "Q3 (Active)": round(Q3, 2),
            "IQR": round(IQR, 2),
            "Upper Threshold": round(upper_bound, 2),
            "Outlier Count": total_outliers,
            "Outlier %": round(pct_outliers, 2),
        }
    )

outlier_df = pd.DataFrame(outlier_summary)
print("--- ACTIVE-RECORD IQR OUTLIER SUMMARY ---")
print(outlier_df)

#  Generate Boxplots with Threshold Overlay
plt.figure(figsize=(16, 9))
sns.set_theme(style="whitegrid")

for i, col in enumerate(outlier_cols_production, 1):
    plt.subplot(2, 3, i)

    active_data = production[production[col] > 0][col]

    # Calculate upper threshold dynamically for plot
    Q1 = active_data.quantile(0.25)
    Q3 = active_data.quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + (1.5 * IQR)

    # Boxplot of active production
    sns.boxplot(y=active_data, color="skyblue")

    # Add red dashed line for upper bound threshold
    plt.axhline(
        upper_bound,
        color="red",
        linestyle="--",
        linewidth=1.5,
        label=f"Upper Bound ({upper_bound:,.1f})",
    )

    plt.title(f"Boxplot: {col} (Active)", fontsize=11, fontweight="bold")
    plt.ylabel("Value")
    plt.legend(loc="upper right", fontsize=9)

plt.tight_layout()
plt.show()
```

 Outlier Treatment can be done using
</h1>

<ul style="margin-top: 12px; line-height: 1.6; color: #333333; padding-left: 24px;">
  <li style="margin-bottom: 10px;">
    <strong><code>{col}_capped</code>:</strong> Values exceeding the 99th percentile of active records are clipped to <strong>P<sub>99</sub></strong>. This retains the full sample size while suppressing extreme outlier spikes for model stability.
  </li>
  <li style="margin-bottom: 10px;">
    <strong><code>{col}_log</code>:</strong> Calculates <strong>ln(1 + x)</strong>. Compresses the long right-hand tail into a normal/Gaussian distribution while preserving <code>0</code> production days as <code>0</code>.
  </li>
  <li style="margin-bottom: 10px;">
    <strong><code>{col}_capped_log</code>:</strong> Combines 99th percentile capping with logarithmic transformation for maximum variance compression in linear models.

```python
# Define target production columns
outlier_cols_production = [
    "oil_rate_bpd",
    "gas_rate_mcfpd",
    "oil_equivalent_rate_bpd",
    "water_rate_bpd",
]

# Create a copy of the dataset for feature engineering
production_fe = production.copy()

# Initialize a tracking summary
transformation_summary = []

for col in outlier_cols_production:
    # ---------------------------------------------------------
    #  Apply Log Transformation: log1p(x) = log(1 + x)
    # ---------------------------------------------------------
    # Handles zeros smoothly: log(1 + 0) = 0
    production_fe[f"{col}_log"] = np.log1p(production_fe[col])

    transformation_summary.append(
        {
            "Feature": col,
            "Raw Max": round(production_fe[col].max(), 2),
            "Log Transformed Max": round(
                production_fe[f"{col}_log"].max(), 2
            )
        }
    )

# Display transformation results summary table
summary_df = pd.DataFrame(transformation_summary)
print("--- FEATURE ENGINEERING SUMMARY ---")
print(summary_df)
```

  Creating Derived column for production
</h2>

```python
# Oil Equivalent Rate expressed in Barrels Per Day (usually abbreviated as BOE/D or BOEPD),
# it represents the total daily production of all combined energy commodities.

# 'oil_rate_bpd' stands for Oil Production Rate in Barrels Per Day.
# Gas_rate_mcfpd stands for Gas Rate in Thousand Cubic Feet(mcf) Per Day(pd).

# 'oil_PE_per_well'= 'oil_rate_bpd' / 'producing_wells'
# 'gas_PE_per_well' = 'gas_rate_mcfpd' / 'producing_wells'
```

```python
### Production Efficiency per well
# Instead of only looking at total oil production,
# calculate oil production per producing well & gas production per producing well.

# Replace 0 wells with NaN to prevent division by zero, then fillna(0) no production efficiency for shut-ins
wells_non_zero = production_fe["producing_wells"].replace(0, np.nan)

# Production Efficiency per Well for oil and gas:
production_fe["oil_PE_per_well"] = (production_fe["oil_rate_bpd"] / wells_non_zero).fillna(0)
production_fe["gas_PE_per_well"] = (production_fe["gas_rate_mcfpd"] / wells_non_zero).fillna(0)

# Log-transformed per-well efficiencies
production_fe["oil_PE_per_well_log"] = np.log1p(production_fe["oil_PE_per_well"])
production_fe["gas_PE_per_well_log"] = np.log1p(production_fe["gas_PE_per_well"]
)

# Phase Behavior Ratios (WOR & GOR) ---
# Replace zero oil rate with NaN to avoid floating-point infinity (inf)
oil_non_zero = production_fe["oil_rate_bpd"].replace(0, np.nan)

# Water-to-Oil Ratio
production_fe["water_oil_ratio"] = (production_fe["water_rate_bpd"] / oil_non_zero).fillna(0)

# Gas-to-Oil Ratio (Standard O&G formula: Gas MCFPD * 1000 / Oil BPD)
production_fe["gas_oil_ratio"] = ((production_fe["gas_rate_mcfpd"] * 1000) / oil_non_zero).fillna(0)

# Log-transform fluid ratios to stabilize variance
production_fe["water_oil_ratio_log"] = np.log1p(production_fe["water_oil_ratio"])
production_fe["gas_oil_ratio_log"] = np.log1p(production_fe["gas_oil_ratio"])

### Production year
# extracting the year from production_month
production_fe["production_year"] = (production_fe["production_month"].dt.year)

### Production Month Number
# Extracting the month from production_month
production_fe["production_month_number"] = (production_fe["production_month"].dt.month)

### Production duration
# calculate the time from installation to production month
production_fe["production_age_years"] = (production_fe["production_month"] - production_fe["install_date"]).dt.days /365.25
```

```python
### Lease_stage: Active/Removed/New_lease
# Define the conditions
conditions = [
    # Condition 1: Active lease/platform, but no producing wells (Development/Preparation)
    (production_fe['production_month'].notna()) & 
    (production_fe['removal_date'].isna()) & 
    (production_fe['producing_wells'] == 0),
    
    # Condition 2: Active lease/platform with active production
    (production_fe['production_month'].notna()) & 
    (production_fe['removal_date'].isna()) & 
    (production_fe['producing_wells'] > 0),
    
    # Condition 3: Decommissioned / Removed asset
    (production_fe['removal_date'].notna())
]

# Assign labels based on conditions
choices = [
    'New Lease / Platform Prep', 
    'Active Producing', 
    'Decommissioned / Removed'
]

# Create the derived column
production_fe['lease_stage'] = np.select(conditions, choices, default='Unknown')
```

  Creating Derived column for platform

```python
#To find the plaform age, we have install_date and removal_date which gives the operational_age
#but there are platforms which are still in production/active,
# to calculate the platform age in production/active status,
# we fix a date till 31/12/2025, because the datas of production,incs are filtered from 01/01/2020 to 31/12/2025.
# so we are calculating the platform_current_age until 2025 as per data.
```

```python
# Reference benchmark cutoff matching your dataset range
analysis_date = pd.Timestamp("2025-12-31")

### Platform status
platform["platform_status"] = np.where(platform["removal_date"].isna(),"Active/Not Removed","Removed")

# Active Platform Age (as of 2025-12-31 benchmark)
platform["current_age_years"] = np.where(
    platform["removal_date"].isna(),
    (analysis_date - platform["install_date"]).dt.days / 365.25,
    np.nan
)

### Operational duration:Total Operational Lifespan for Decommissioned Platforms
platform["operational_years"] = (
    platform["removal_date"] -
    platform["install_date"]
).dt.days / 365.25

### Water-depth category
# Numerical variable --> pd.cut() --> Categorical variable. This is called binning or discretization.
platform["water_depth_category"] = pd.cut(
    platform["water_depth_ft"],
    bins=[-np.inf, 100, 500, 1000, np.inf],
    labels=["Shallow", "Moderate", "Deep", "Very Deep"]
)

### Coordinate availability
platform["coordinates_available"] = np.where(
    platform["latitude"].notna() &
    platform["longitude"].notna(),
    "Available",
    "Missing"
)
```

  Creating Derived column for incs

```python
# Ensure individual INC columns have missing values filled with 0
warning_inc = incs["warning_INCs"].fillna(0)
comp_shutin_inc = incs["component_shut-in_INCs"].fillna(0)
fac_shutin_inc = incs["facility_Shut-in_INCs"].fillna(0)

# Total overall incidents
incs["total_incidents"] = warning_inc + comp_shutin_inc + fac_shutin_inc

# Total shut-in incidents (Component + Facility level severity)
incs["total_shut_in_INCs"] = comp_shutin_inc + fac_shutin_inc


### Incidents per inspection
# .replace(): This method is used to replace specified values in a DataFrame or Series.
# 0: The value that we want to replace.
# np.nan: The value that will replace the specified value (in this case, NaN from the NumPy library).

# Incidents per Inspection (Handling zero inspection records)
inspections_non_zero = incs["inspection_records"].replace(0, np.nan)

incs["incidents_per_inspection"] = (
    incs["total_incidents"] / inspections_non_zero
).fillna(0)

# Warning Percentage of Total Incidents
incidents_non_zero = incs["total_incidents"].replace(0, np.nan)

incs["warning_incident_pct"] = (
    (warning_inc / incidents_non_zero) * 100
).fillna(0)
```

### Interperation:
** removal_date_x vs removal_date_y (Data Leakage / Merge Artifacts):
Having both _x and _y indicates a pd.merge() created duplicate columns because removal_date existed in both datasets.
removal_date_y matches operational_years perfectly (89.98% missing),suggesting removal_date_y came from the table containing historical removal records.

You should combine (coalesce) these two columns into a single removal_date and drop the duplicates.

** High Missingness in Ratios (~51% – 55%):
water_oil_ratio, gas_oil_ratio, oil_PE_per_well, and gas_PE_per_well are missing in ~33,000–35,000 rows. In oil & gas datasets,
this almost always happens due to:

* Division by zero (e.g., zero producing wells or zero oil production).
* Inactive / prep-stage leases where production rates are zero.

```python
# Consolidate Duplicate Removal Dates:
# Combine removal_date_x and removal_date_y into one unified removal_date column,filling missing values from one with the other.

# Coalesce removal dates from merge duplicates
production_platform_analysis['removal_date'] = (
    production_platform_analysis ['removal_date_x']
    .fillna(production_platform_analysis['removal_date_y'])
)

# Drop redundant merge columns
production_platform_analysis.drop(columns=['removal_date_x', 'removal_date_y'], inplace=True)



# Handle Ratios & Per-Well Production Metrics:
# For active leases that have zero producing wells or zero oil production, set ratios safely to 0.0 instead of letting them default to NaN or Inf.



# 1. Per-Well Metrics (Safely assign 0 if producing_wells == 0)
production_platform_analysis['oil_PE_per_well'] = np.where(
    production_platform_analysis['producing_wells'] > 0,
    production_platform_analysis['oil_rate_bpd'] / production_platform_analysis['producing_wells'],
    0.0
)

production_platform_analysis['gas_PE_per_well'] = np.where(
    production_platform_analysis['producing_wells'] > 0,
    production_platform_analysis['gas_rate_mcfpd'] / production_platform_analysis['producing_wells'],
    0.0
)

# 2. Water-Oil Ratio (WOR) & Gas-Oil Ratio (GOR)
production_platform_analysis['water_oil_ratio'] = np.where(
    production_platform_analysis['oil_rate_bpd'] > 0,
    production_platform_analysis['water_rate_bpd'] / production_platform_analysis['oil_rate_bpd'],
    0.0
)

production_platform_analysis['gas_oil_ratio'] = np.where(
    production_platform_analysis['oil_rate_bpd'] > 0,
    production_platform_analysis['gas_rate_mcfpd'] / production_platform_analysis['oil_rate_bpd'],
    0.0
)

# Explicit Handling of Operational Dates
# For active platforms where site_clear_date and removal_date are expected to be blank,
# keep them as missing (NaT) or create an explicit flag so downstream machine learning models do not drop these rows.


# Flag indicating active assets (no removal date)
production_platform_analysis['is_active_asset'] = production_platform_analysis['removal_date'].isna()

# Recalculate operational years if removal_date is present, otherwise standard age
production_platform_analysis['operational_years'] = production_platform_analysis['operational_years'].fillna(
    production_platform_analysis['current_age_years']
)
```

❓ Question:   Did production increase or decrease over time?
</h1>

```python
# =============================================
# MULTIVARIATE: Time Series Aggregation:
# =============================================

# Group by month and SORT chronologically
monthly_trend = (
    production_fe.groupby("production_month")
    .agg({
        "oil_rate_bpd": "sum",
        "gas_rate_mcfpd": "sum",
        "producing_wells": "sum"
    })
    .reset_index()
    .sort_values("production_month")  # Crucial for valid iloc trajectory
)
# Calculate trajectory metrics (Oil)
first_month_oil = monthly_trend["oil_rate_bpd"].iloc[0]
last_month_oil = monthly_trend["oil_rate_bpd"].iloc[-1]
pct_change_oil = ((last_month_oil - first_month_oil) / first_month_oil) * 100 

# Calculate trajectory metrics (Gas)
first_month_gas = monthly_trend["gas_rate_mcfpd"].iloc[0]
last_month_gas = monthly_trend["gas_rate_mcfpd"].iloc[-1]
pct_change_gas = ((last_month_gas - first_month_gas) / first_month_gas) * 100 
print("==================================================")
print("--- PRODUCTION TRAJECTORY EVALUATION ---")
print(f"Initial Oil Rate: {first_month_oil:,.2f} BPD | Final Oil Rate: {last_month_oil:,.2f} BPD")
print(f"Overall Oil Trend Shift: {pct_change_oil:+.2f}%")
print(f"Initial Gas Rate: {first_month_gas:,.2f} MCFPD | Final Gas Rate: {last_month_gas:,.2f} MCFPD")
print(f"Overall Gas Trend Shift: {pct_change_gas:+.2f}%")
print("==================================================")
```

```python
# Compute calendar days in each production month
production_platform_analysis['days_in_month'] = production_platform_analysis['production_month'].dt.days_in_month

# Calculate total monthly oil volume (BPD * Days in Month)
production_platform_analysis['oil_volume_bbl'] = production_platform_analysis['oil_rate_bpd'] * production_platform_analysis['days_in_month']

# Compute key totals
total_oil_bbl = production_platform_analysis['oil_volume_bbl'].sum()
total_oil_mmbbl = total_oil_bbl / 1e6

print("==================================================")
print(f" Total Cumulative Oil Produced: {total_oil_bbl:,.2f} Barrels")
print(f" Total Cumulative Oil Produced: {total_oil_mmbbl:,.2f} Million Barrels (MMbbl)")
print("==================================================")
```

```python
# ================================================================
#            UNIVARIATE TIME SERIES ANALYSIS
# ================================================================

# Group by production month to get monthly aggregated oil volume (MMbbl)
monthly_trend = (
    production_platform_analysis
    .groupby('production_month')['oil_volume_bbl']
    .sum()
    .reset_index()
)
monthly_trend['oil_volume_mmbbl'] = monthly_trend['oil_volume_bbl'] / 1e6

# Calculate overall total for display
total_oil_mmbbl = monthly_trend['oil_volume_mmbbl'].sum()

# Create time series line chart
fig = px.line(
    monthly_trend,
    x='production_month',
    y='oil_volume_mmbbl',
    title=f"Monthly Oil Production Trend<br><sup>Total Cumulative Production: <b>{total_oil_mmbbl:,.2f} MMbbl</b></sup>",
    labels={'production_month': 'Production Month', 'oil_volume_mmbbl': 'Oil Volume (MMbbl)'}
)

# Style line and hover layout
fig.update_traces(
    line=dict(color="steelblue", width=2.5),
    hovertemplate="<b>Date:</b> %{x|%b %Y}<br><b>Volume:</b> %{y:,.2f} MMbbl<extra></extra>"
)


fig.show()
```

```python
# ================================================
#             BIVARIATE ANALYSIS
# ================================================

# Calculate correlation between operational age and daily production
age_corr = production_platform_analysis["operational_years"].corr(
    production_platform_analysis["oil_equivalent_rate_bpd"]
)
print(
    f"Correlation between Platform Age (Years) and Daily BOE: {age_corr:.2f}"
)

# Bin platform operational years into age brackets for comparative grouping
age_bins = [0, 10, 20, 30, 40, 100]
age_labels = ["0-10 Yrs", "11-20 Yrs", "21-30 Yrs", "31-40 Yrs", "40+ Yrs"]

production_platform_analysis["age_bracket"] = pd.cut(
    production_platform_analysis["operational_years"],
    bins=age_bins,
    labels=age_labels,
    right=True,
)

# Aggregate production metrics across age brackets
age_summary = (
    production_platform_analysis.groupby("age_bracket", observed=True)
    .agg(
        avg_boe_bpd=("oil_equivalent_rate_bpd", "mean"),
        median_boe_bpd=("oil_equivalent_rate_bpd", "median"),
        platform_records=("oil_equivalent_rate_bpd", "count"),
    )
    .reset_index()
)

print(age_summary)

# ==========================================================================================
# VISULIZATION: BIVARIATE ANALYSIS(Based on oil_equivalent_rate_bpd & Platform Age Brackets)
# ===========================================================================================
plt.figure(figsize=(9, 5))
sns.barplot(
    data=age_summary,
    x="age_bracket",
    y="avg_boe_bpd",
    hue="age_bracket",
    palette="Blues_d",
    legend=False)

plt.title(
    "Average Production Performance (BOE/D) by Platform Age Category",
    fontsize=13,
    weight="bold",
)
plt.xlabel("Platform Operational Age Bracket", fontsize=11)
plt.ylabel("Mean Oil Equivalent Rate (BPD)", fontsize=11)
plt.tight_layout()
plt.show()
```

<h1 style="color: #FFFFFF; background: linear-gradient(135deg, #4F46E5, #7C3AED); padding: 14px; border-radius: 10px; text-align: left;">
  ❓ Question:  Did production increase or decrease over time in Gulf of America?(Trend Analysis)
</h1> 

```python
#Time Series Aggregation: Did production increase or decrease?
# ==============================================================================
# Group by month and SORT chronologically
monthly_trend = (
    production_platform_analysis.groupby("production_month")
    .agg({
        "oil_rate_bpd": "sum",
        "gas_rate_mcfpd": "sum",
        "producing_wells": "sum"
    })
    .reset_index()
    .sort_values("production_month")  # Crucial for valid iloc trajectory
)
# Calculate trajectory metrics (Oil)
first_month_oil = monthly_trend["oil_rate_bpd"].iloc[0]
last_month_oil = monthly_trend["oil_rate_bpd"].iloc[-1]
pct_change_oil = ((last_month_oil - first_month_oil) / first_month_oil) * 100 

# Calculate trajectory metrics (Gas)
first_month_gas = monthly_trend["gas_rate_mcfpd"].iloc[0]
last_month_gas = monthly_trend["gas_rate_mcfpd"].iloc[-1]
pct_change_gas = ((last_month_gas - first_month_gas) / first_month_gas) * 100 
print("==================================================")
print("--- PRODUCTION TRAJECTORY EVALUATION ---")
print(f"Initial Oil Rate: {first_month_oil:,.2f} BPD | Final Oil Rate: {last_month_oil:,.2f} BPD")
print(f"Overall Oil Trend Shift: {pct_change_oil:+.2f}%")
print(f"Initial Gas Rate: {first_month_gas:,.2f} MCFPD | Final Gas Rate: {last_month_gas:,.2f} MCFPD")
print(f"Overall Gas Trend Shift: {pct_change_gas:+.2f}%")
print("==================================================")
```

```python
# MULTIVARIATE TIME-SERIES ANALYSIS
# ==============================================================================
# DUAL-AXIS TRAJECTORY VISUALIZATION 
# ==============================================================================
plt.figure(figsize=(12, 6))

x = monthly_trend["production_month"]
y1 = monthly_trend["oil_rate_bpd"]
y2 = monthly_trend["gas_rate_mcfpd"]

# --- LEFT Y-AXIS: Oil Rate ---
color_oil = '#1f77b4'  # Blue
plt.plot(x, y1, color=color_oil, linewidth=2.0, label="Oil Rate (BPD)")
plt.xlabel("Production Month", fontsize=12, fontweight="bold", labelpad=10)
plt.ylabel("Oil Rate (BPD)", color=color_oil, fontsize=12, fontweight="bold")
plt.tick_params(axis='y', labelcolor=color_oil)
plt.grid(True, linestyle="--", alpha=0.5)

# --- SWITCH TO RIGHT Y-AXIS ---
plt.twinx()

# --- RIGHT Y-AXIS: Gas Rate ---
color_gas = '#2ca02c'  # Green
plt.plot(x, y2, color=color_gas, linewidth=2.0, linestyle="--", label="Gas Rate (MCFPD)")
plt.ylabel("Gas Rate (MCFPD)", color=color_gas, fontsize=12, fontweight="bold")
plt.tick_params(axis='y', labelcolor=color_gas)

# Formatting
plt.title("Historical Oil & Gas Production Trajectory (production_platform_analysis)", fontsize=15, fontweight="bold", pad=15)
plt.tight_layout()

plt.show()
```
